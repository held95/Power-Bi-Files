```python
import fitz  # PyMuPDF
import re
from openpyxl import Workbook

# Abrir o PDF
caminho_pdf = "/Conta 3 IRMAOS - Mar.pdf"  # Altere o caminho se necessário
doc = fitz.open(caminho_pdf)

# Expressões para encontrar os valores
padrao_uc = r"UC\s*[:\-]?\s*(\d+)"
padrao_compensacao = r"compensação\s*[:\-]?\s*([\d,\.]+)"
padrao_consumo = r"consumo\s*[:\-]?\s*([\d,\.]+)"
padrao_saldo = r"saldo\s*[:\-]?\s*([\d,\.\-]+)"

# Lista para armazenar os dados
dados_extraidos = []

# Extrair texto de cada página
for pagina in doc:
    texto = pagina.get_text()

    ucs = re.findall(padrao_uc, texto, re.IGNORECASE)
    compensacoes = re.findall(padrao_compensacao, texto, re.IGNORECASE)
    consumos = re.findall(padrao_consumo, texto, re.IGNORECASE)
    saldos = re.findall(padrao_saldo, texto, re.IGNORECASE)

    # Combinar os dados de forma segura
    for i in range(min(len(ucs), len(compensacoes), len(consumos), len(saldos))):
        dados_extraidos.append([
            compensacoes[i],
            consumos[i],
            saldos[i],
            ucs[i]
        ])

# Criar e salvar a planilha
wb = Workbook()
ws = wb.active
ws.title = "Dados"

# Cabeçalhos
ws.append(["Compensação", "Consumo", "Saldo", "UC"])

# Adicionar os dados extraídos
for linha in dados_extraidos:
    ws.append(linha)

# Salvar o arquivo
wb.save("dados_extraidos(2).xlsx")
print("Planilha criada com sucesso!")
```

```python
import fitz  # PyMuPDF
import re
import os
from openpyxl import Workbook

# Caminho da pasta com os PDFs
pasta_pdfs = "/mnt"

# Criar planilha
wb = Workbook()
ws = wb.active
ws.title = "Dados"
ws.append(["Arquivo", "Compensação", "Consumo", "Saldo", "UC"])

# Expressões ajustadas com maior flexibilidade
padrao_uc = r"\b(\d{10})\b"
padrao_consumo = r"Consumo\s*-\s*TE\s*[A-Z]{3}/\d{2}.*?([\d.]+)"
padrao_compensacao = r"Energ\s+Atv\s+Inj\.\s+oUC\s+mPT\s*-\s*TUSD\s*[A-Z]{3}/\d{2}.*?([\d.]+)"
padrao_saldo = r"Saldo em Energia.*?:.*?([\d.,]+)"

# Processar todos os PDFs da pasta
for nome_arquivo in os.listdir(pasta_pdfs):
    if nome_arquivo.lower().endswith(".pdf"):
        caminho_pdf = os.path.join(pasta_pdfs, nome_arquivo)
        try:
            doc = fitz.open(caminho_pdf)
            texto = ""
            for pagina in doc:
                texto += pagina.get_text()

            # Aplicar regex
            uc = re.search(padrao_uc, texto)
            consumo = re.search(padrao_consumo, texto, re.IGNORECASE | re.DOTALL)
            compensacao = re.search(padrao_compensacao, texto, re.IGNORECASE | re.DOTALL)
            saldo = re.search(padrao_saldo, texto, re.IGNORECASE)

            # Extrair valores
            uc_val = uc.group(1) if uc else ""
            consumo_val = consumo.group(1) if consumo else ""
            compensacao_val = compensacao.group(1) if compensacao else ""
            saldo_val = saldo.group(1) if saldo else ""

            # Adicionar à planilha
            ws.append([nome_arquivo, compensacao_val, consumo_val, saldo_val, uc_val])

        except Exception as e:
            print(f"Erro ao processar {nome_arquivo}: {e}")

# Salvar planilha
caminho_saida = os.path.join(pasta_pdfs, "dados_extraidos_todos(3).xlsx")
wb.save(caminho_saida)
print("✔ Planilha final criada com sucesso!")
```

```python
import pandas as pd
import xml.etree.ElementTree as ET
import os
from glob import glob

# Caminho da pasta onde estão os arquivos
pasta_arquivos = '/mnt/Dados_usinas_xml'

# Pegar todos os arquivos .txt da pasta
arquivos_txt = glob(os.path.join(pasta_arquivos, '*.txt'))

# Lista para armazenar os dados extraídos
dados = []

# Loop por todos os arquivos
for arquivo in arquivos_txt:
    try:
        tree = ET.parse(arquivo)
        root = tree.getroot()

        nome_arquivo = os.path.basename(arquivo)  # Nome do arquivo, ex: relatorio_beto_usina.txt

        for linha in root.findall("Linha"):
            dado = {
                'ucs': linha.findtext("Instalacao"),
                'compensacao': float(linha.findtext("Qtd_compensacao", default="0").replace(",", ".")),
                'consumo': float(linha.findtext("Qtd_consumo", default="0").replace(",", ".")),
                'mês referente': linha.findtext("Periodo_uso"),
                'saldo': float(linha.findtext("Qtd_saldo_atual", default="0").replace(",", ".")),
                'arquivo_origem': nome_arquivo
            }
            dados.append(dado)

    except Exception as e:
        print(f"Erro ao processar {arquivo}: {e}")

# Criar DataFrame
df = pd.DataFrame(dados)

# Salvar o resultado no Excel
output_path = os.path.join(pasta_arquivos, 'relatorio_compilado.xlsx')
df.to_excel(output_path, index=False)

print(f"✅ Arquivo Excel gerado com sucesso em: {output_path}")
```
